```python
import numpy as np
import matplotlib.pyplot as plt
import os

file_name = "values.txt"

# =========================== Stage 1: Hardcoded Input ===========================

hardcoded_values = [
    "0 1 5 -10 10",  # y = x + 5 (Straight line)
    "0 0 7 -10 10",  # y = 7 (Horizontal line)
    "0 -2 3 -10 10"  # y = -2x + 3 (Straight line)
]

print("\n=== Stage 1: Hardcoded Input ===")
for i, line in enumerate(hardcoded_values):
    data = list(map(float, line.split()))
    a, b, c, x_start, x_end = data
    x = np.linspace(int(x_start), int(x_end), 100)
    y = a * x**2 + b * x + c

    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label=f"Hardcoded Set {i+1}", color="purple")
    plt.title(f"Weather Modeling - Stage 1: Hardcoded Input (Set {i+1})")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.legend()
    plt.grid(True)
    plt.savefig(f"weather_model_hardcoded_set_{i+1}.png")
    plt.show()

# =========================== Stage 2: Keyboard Input ===========================

print("\n=== Stage 2: Keyboard Input ===")
num_inputs = int(input("How many sets of values do you want to enter? "))

keyboard_values = []
for i in range(num_inputs):
    user_input = input(f"Enter values for set {i+1} (format: a b c x_start x_end): ")
    keyboard_values.append(user_input)

for i, line in enumerate(keyboard_values):
    data = list(map(float, line.split()))
    a, b, c, x_start, x_end = data
    x = np.linspace(int(x_start), int(x_end), 100)
    y = a * x**2 + b * x + c

    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label=f"Keyboard Input Set {i+1}", color="blue")
    plt.title(f"Weather Modeling - Stage 2: Keyboard Input (Set {i+1})")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.legend()
    plt.grid(True)
    plt.savefig(f"weather_model_keyboard_set_{i+1}.png")
    plt.show()

# =========================== Stage 3: File Input ===========================

# Check if the file exists, otherwise create it with hardcoded values
if not os.path.exists(file_name) or os.stat(file_name).st_size == 0:
    with open(file_name, 'w') as file:
        for line in hardcoded_values:
            file.write(line + "\n")

# Ask user if they want to add more custom input to the file
choice = input("\nDo you want to add custom values to the file? (yes/no): ").strip().lower()
if choice == "yes":
    num_file_inputs = int(input("How many sets of values do you want to add to the file? "))

    with open(file_name, 'a') as file:
        for i in range(num_file_inputs):
            user_input = input(f"Enter values for set {i+1} (format: a b c x_start x_end): ")
            file.write(user_input + "\n")

# Read from file and generate graphs
print("\n=== Stage 3: File Input ===")
with open(file_name, 'r') as file:
    lines = file.readlines()

for i, line in enumerate(lines):
    data = line.strip().split()

    if len(data) != 5:
        print(f"Skipping invalid input at line {i+1}: {data}")
        continue

    try:
        a, b, c, x_start, x_end = map(float, data)
        x_start = int(x_start)
        x_end = int(x_end)

        x = np.linspace(x_start, x_end, 100)
        y = a * x**2 + b * x + c

        plt.figure(figsize=(10, 6))
        plt.plot(x, y, label=f"File Input Set {i+1}", color="green")
        plt.title(f"Weather Modeling - Stage 3: File Input (Set {i+1})")
        plt.xlabel("X")
        plt.ylabel("Y")
        plt.legend()
        plt.grid(True)
        plt.savefig(f"weather_model_file_set_{i+1}.png")
        plt.show()

    except ValueError:
        print(f"Skipping invalid input at line {i+1}: {data}")
```
